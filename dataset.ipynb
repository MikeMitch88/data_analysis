```python
# Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris(as_frame=True)
df = iris.frame

# Display first 5 rows
df.head()
```

```python
# Check data types and missing values
df.info()

# Check for missing values
df.isnull().sum()
```

```python
# handling  missing values (for other datasets):
df.dropna(inplace=True)  

```

```python
# TASK 2

# Basic statistics
df.describe()
```

```python
# Group by species and compute the mean of each numeric column
df.groupby('target').mean()

# Map target numbers to species names (for better readability)
df['species'] = df['target'].map(dict(zip(range(3), iris.target_names)))
df.groupby('species').mean()
```

```python
# Interesting pattern 
# Check which species has the longest average petal length
df.groupby('species')['petal length (cm)'].mean()
```

```python
#TASK 3
#Line Chart
plt.figure(figsize=(10, 5))
plt.plot(df.index, df['sepal length (cm)'], label='Sepal Length')
plt.plot(df.index, df['petal length (cm)'], label='Petal Length')
plt.title('Trend of Sepal and Petal Length')
plt.xlabel('Index')
plt.ylabel('Length (cm)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Bar Chart

df.groupby('species')['petal length (cm)'].mean().plot(kind='bar', color='skyblue')
plt.title('Average Petal Length by Species')
plt.ylabel('Petal Length (cm)')
plt.xticks(rotation=0)
plt.show()
```

```python
#Histogram

df['sepal length (cm)'].plot(kind='hist', bins=15, color='orange', edgecolor='black')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.show()
```

```python
#Scatter Plot – Sepal length vs. petal length

sns.scatterplot(data=df, x='sepal length (cm)', y='petal length (cm)', hue='species')
plt.title('Sepal Length vs. Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.show()
```

```python
#Findings and Observations

- Iris-virginica has the longest average petal length.
- The histogram shows that most sepal lengths are between 5 and 6 cm.
- Scatter plot reveals clear separation between species, useful for classification.
- Line chart shows how sepal and petal lengths vary across samples (indexes).
```
